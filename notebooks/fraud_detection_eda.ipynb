# FraudGuard AI — Exploratory Data Analysis

Quick EDA on `data/creditcard.csv` (real Kaggle dataset or the synthetic one from `generate_sample_data.py`).

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

df = pd.read_csv("../data/creditcard.csv")
df.shape
```

```python
df.head()
```

```python
df.info()
df.isna().sum().sum()
```

## Class imbalance

```python
class_counts = df["Class"].value_counts()
fraud_rate = df["Class"].mean() * 100
print(class_counts)
print(f"Fraud rate: {fraud_rate:.3f}%")

class_counts.plot(kind="bar", color=["#2ecc71", "#e74c3c"])
plt.title("Class Distribution (0 = Legit, 1 = Fraud)")
plt.ylabel("Count")
plt.show()
```

## Amount distribution: fraud vs legit

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
sns.histplot(df[df["Class"] == 0]["Amount"], bins=50, ax=ax[0], color="#2ecc71")
ax[0].set_title("Legit Transaction Amounts")
sns.histplot(df[df["Class"] == 1]["Amount"], bins=50, ax=ax[1], color="#e74c3c")
ax[1].set_title("Fraud Transaction Amounts")
plt.tight_layout()
plt.show()
```

## Time-of-day pattern

```python
df["hour"] = (df["Time"] % 86400) // 3600

fig, ax = plt.subplots(figsize=(10, 4))
sns.histplot(df[df["Class"] == 0]["hour"], bins=24, color="#2ecc71", label="Legit", stat="density", alpha=0.5)
sns.histplot(df[df["Class"] == 1]["hour"], bins=24, color="#e74c3c", label="Fraud", stat="density", alpha=0.5)
plt.legend()
plt.title("Transaction Hour of Day: Fraud vs Legit")
plt.xlabel("Hour")
plt.show()
```

## Correlation of PCA features with Class

```python
v_cols = [c for c in df.columns if c.startswith("V")]
corr_with_class = df[v_cols + ["Class"]].corr()["Class"].drop("Class").sort_values()

corr_with_class.plot(kind="barh", figsize=(6, 8), color="steelblue")
plt.title("Correlation of V1-V28 with Class")
plt.tight_layout()
plt.show()
```
